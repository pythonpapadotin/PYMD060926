```python
def check_if_all_small_alphabets_are_present(sentence):
    attendance_register = []

    for ascii_value in range(97, 122+1):
        attendance_register.append(None)

    for ch in sentence:
        index_position_in_register = ord(ch) - 97
        if index_position_in_register >= 0 and index_position_in_register <= 25:
            attendance_register[index_position_in_register] = True

    if all(attendance_register) is True:
        print('Yes all the alphabets are present!')
    else:
        print('No, All the alphabets are not present!')
```

```python
# Inheritance
class Organism:
    def breathe(self):
        print(f'{self.name} is breathing!')

class Animal(Organism):
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} eat in Animal!")

class Cat(Animal):
    def speak(self):
        print(f"{self.name} Meow!")


class Dog(Animal):
    def speak(self):
        print(f"{self.name} Woof!")

    def eat(self):
        print(f"{self.name} eat in Dog!")
```

```python
yuki = Dog('Yuki Singh')
```

```python
yuki.eat()
```

```python
yuki.breathe()
```

```python
# Polymorphism

for animal in [Dog('Yuki Singh'), Cat('Duster')]:
    animal.speak()
```

```python
class BankAccount:
    def __init__(self, initial_balance, name_of_person):
        self.balance = initial_balance  # balance is an object attribute/property
        self.name = name_of_person # name is an object attribute/property

    def get_balance(self):
        return self.balance

    def get_name(self):
        return self.name
```

```python
class BankAccount:
    def __init__(self, initial_balance, name_of_person):
        self.__balance = initial_balance  # __balance is an object attribute/property
        self.__name = name_of_person # __name is an object attribute/property

    def get_balance(self):
        return self.__balance

    def get_name(self):
        return self.__name

    def set_balance(self, value):
        if value < 0:
            raise Exception("Negative Value not allowed!")
        self.__balance = value
```

```python
mangesh_account = BankAccount(1000000000000, "Mangesh")
```

```python
mangesh_account.get_balance()
```
