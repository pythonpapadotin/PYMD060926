```python
# dict

key: value pairs

x = {} curly brackets

-> unordred : There is no suppport index or indexing.
-> keys are unique in nature , you can not repeat a key.
-> keys are immutable data type (They can be int, float, tuple, bool, str, None) but they can not be (list, dict, set)
-> you can access any value in O(1) or constant time, if you know the key. Fast Speed.


Mutable/Editable:-
list, dict, set

Immutable/Non Editable:-
int, float, tuple, bool, str, None
```

```python
set_a = {'a', 'b', 'c'}

set_b = {'d', 'a', 'b'}

set_a = set_a.union(set_b)

print(set_a)
```

```python
set_a = {'a', 'b', 'c'}

set_b = {'d', 'a', 'b'}

set_a = set_a.intersection(set_b)

print(set_a)
```
