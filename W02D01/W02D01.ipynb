```python
def custom_number_printing(input_list):
    for item in input_list:
        if type(item) == int and item%2 != 0:
            print(f'odd {item}')
        elif item is None:
            print('null')
        else:
            print(item)
```

```python
custom_number_printing([None, None, 67, 79, 88])
```

```python
custom_number_printing(['happy', 'mango', 'ice', None, 45, 76])
```

```python
# while loops

while condition:
    code 
    code

code will run again and again in a loop unless the condition becomes False.
code will only run if the condition is True
```

```python
count = 5

while count > 10:
    print('Counting more than 10')
```

```python
count = 5

while count > 0:
    print(f'Positive no {count}')
    count -= 1

print(f'When the loop ends, the final value of count is {count}')
```

```python
# break and continue

# break

# you can break the loop anytime inside the loop even if the condition is True. You will execute the next lines which are present.

count = 5

while count > 0:
    print(f'Positive no {count}')
    count -= 1
    if count == 3:
        break

print(f'When the loop ends, the final value of count is {count}')
```

```python

count = 5

while count > 0:
    print(f'Positive no {count}')
    count -= 1
    break

print(f'When the loop ends, the final value of count is {count}')
```

```python
# continue
# skip rest of the line and than continue to the loop condition

youtube_video_sequences = ['video1', 'advertisement', 'video2', 'video3', 'advertisement', 'advertisement']

for video in youtube_video_sequences:
    if video == 'advertisement':
        continue # skip this part, and take on the next one

    print(f'watching {video}')
```

```python
def fun1(first_name, **kwargs):
    rest_of_the_name = ""
    for value in kwargs.values():
        rest_of_the_name += value
        rest_of_the_name += " "
    print(f'You have been approved with a username {first_name} {rest_of_the_name}')
```

```python
fun1("Sadhna")
```

```python
fun1("Mohit", last_name="Chouhan")
```

```python
fun1("Mohit", fathers_name='Vikram', middle_name="Singh", last_name="Chouhan")
```

```python
def fun2(first_name, *args):
    rest_of_the_name = ""
    for value in args:
        rest_of_the_name += value
        rest_of_the_name += " "
    print(f'You have been approved with a username {first_name} {rest_of_the_name}')
```

```python
fun2("Mohit", 'Vikram', "Singh", "Chouhan")
```
