```python
# JSON are very similar to python dictionaries

# 1. None will be null
# 2. '' or any other quote will be converted to double quotes ("")
# 3. True and False will be true and false

student_data = [
    {'name': "Satya", "subject_marks": [99, 89]},
    {'name': "Meenakshi", "subject_marks": [89, 96]},
    {'name': "Tukaram", "subject_marks": [99, 86]}
]
```

```python
import json
actual_json_data = json.dumps(student_data)

print(actual_json_data)
```

```python
for student in student_data:
    print(student.get('name'))
    print(f"Total Marks are {sum(student.get('subject_marks'))}")
```

```python
payment_data = {
    "amount": 89,
    "cgst": 2.4,
    "sgst": 1.8,
    "fees": {'network_fees': 0.89, 'gateway_fees': 0.44},
    "offer": None,
    "advice": False
}
```

```python
import json

converted_json_data = json.dumps(payment_data)

print(converted_json_data)
```

```python
print(payment_data)
```

```python
with open("payment_data.json", "w") as f:
    json.dump(payment_data, f)
```

```python
with open("payment_data.json", "w") as f:
    json.dump(payment_data, f, indent=4)
```

```python
# json.loads()

with open('online_sample.json', 'r') as f:
    converted_python_data_structure = json.load(f)
    print(type(converted_python_data_structure))
```

```python
count = 3
for item in converted_python_data_structure:
    print(item)
    count -= 1
    if count == 0:
        break
    print('\n')
```
